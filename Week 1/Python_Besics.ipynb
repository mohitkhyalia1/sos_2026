#Set

```python
# set are inclosed in {}
# set does not allow duplicate values
set_1 = {1,2,3,5,6,7,5,6,8,'hello'}
print(set_1)
type(set_1)
```

```python
# print(set_1[2])
# 'set' object is not subscriptable
```

Sets are unordered collections, so they use `.add()` instead of `.append()`. Their modification methods return `None` because they update in-place.

```python
print(set_1)
set_2 = set_1.add(9)
#set_2 = set_1.append(10)     'set' object has no attribute 'append'
print(set_2)
print(set_1)
set_3 = set_1
print(set_3)
```

```python
#convert set to list
list_1 = list(set_1)
print(list_1)
```
