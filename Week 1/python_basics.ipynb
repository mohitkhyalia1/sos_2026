## 1. Data Types and Type Checking

One of the first things I noticed while working with datasets was that values are not always stored in the format I expect. Numbers can be read as strings, and simple type mismatches can create confusing errors. Because of that, I spent some time revisiting Python's basic data types.

```python
# The four most common scalar types in Python
age = 21          # int
height = 5.9      # float
name = 'Mohit'    # str
is_student = True # bool

print(type(age), type(height), type(name), type(is_student))
```

```python
# Type conversion — something that trips up a lot of beginners
# A CSV file often reads numbers as strings

value_as_string = '42.7'
print('Original:', value_as_string, '| type:', type(value_as_string))

value_as_float = float(value_as_string)
print('Converted:', value_as_float, '| type:', type(value_as_float))

# int() truncates — it does NOT round
print('int(42.9):', int(42.9))  # gives 42, not 43
```

**Observation:** `int()` truncates toward zero rather than rounding. This matters when converting continuous predictions to integer class labels — use `round()` or a threshold comparison instead.

```python
# isinstance() is more reliable than type() for checks
# because it handles inheritance correctly

x = 5
print(isinstance(x, int))         # True
print(isinstance(x, (int, float))) # True — checks against a tuple of types
print(isinstance(x, float))       # False
```

## 2. Lists — Operations and Comprehensions

Lists are ordered, mutable, and allow duplicates. Most of the time when iterating over data rows or feature values, we use lists.

```python
# Basic list operations
scores = [88, 72, 95, 61, 84, 77, 95, 68]

print('Length:', len(scores))
print('Max:', max(scores))
print('Min:', min(scores))
print('Sum:', sum(scores))
print('Mean:', sum(scores) / len(scores))
print('Sorted:', sorted(scores))           # returns new list
print('Unique count:', len(set(scores)))   # using set to count unique values
```

```python
# Slicing — negative indexing counts from the end
features = ['age', 'income', 'credit_score', 'loan_amount', 'employed']

print('First two features:', features[:2])
print('Last feature:', features[-1])
print('All except last:', features[:-1])
print('Reversed:', features[::-1])
```

```python
# List comprehensions — the pattern used constantly in data processing
# Format: [expression for item in iterable if condition]

raw_values = [1.2, -0.5, 3.1, -2.4, 0.8, -0.1, 4.5]

# Square all values
squared = [v**2 for v in raw_values]
print('Squared:', squared)

# Keep only positive values
positive = [v for v in raw_values if v > 0]
print('Positive only:', positive)

# Replace negatives with 0 (ReLU activation in one line)
relu = [v if v > 0 else 0 for v in raw_values]
print('ReLU output:', relu)
```

**Observation:** The ReLU pattern (`max(0, x)` applied element-wise) is one of the most common activation functions in neural networks. Seeing it expressed as a list comprehension made it feel less abstract.

```python
# Mutability gotcha — lists are passed by reference
# This causes bugs when you think you have a copy but you don't

original = [10, 20, 30]
reference = original      # NOT a copy
actual_copy = original.copy()  # this IS a copy

original.append(40)

print('original:', original)
print('reference (changed too):', reference)
print('actual_copy (unchanged):', actual_copy)
```

**Observation:** This tripped me up when I first tried to preserve the original dataset while modifying a working copy. Always use `.copy()` or `list()` when you need an independent copy.

## 3. Tuples — Immutability and Unpacking

```python
# Tuples are immutable — they cannot be changed after creation
# Common use: returning multiple values from a function,
# or storing things that should not change (like dataset shape)

dataset_shape = (8000, 12)  # 8000 rows, 12 columns
n_rows, n_cols = dataset_shape  # unpacking

print(f'Dataset has {n_rows} rows and {n_cols} columns')

# This is exactly what .shape returns from a NumPy array or Pandas DataFrame
# Learning tuple unpacking here made .shape feel natural later
```

```python
# Swapping values using tuple unpacking — no temp variable needed
a, b = 10, 20
print('Before swap: a =', a, 'b =', b)

a, b = b, a
print('After swap:  a =', a, 'b =', b)
```

## 4. Sets — Uniqueness and Set Operations

```python
# Sets store unique elements only — useful for finding unique classes or labels

predicted_labels = [0, 1, 0, 1, 1, 0, 2, 1, 2, 0]
unique_classes = set(predicted_labels)
print('Unique classes predicted:', unique_classes)
print('Number of classes:', len(unique_classes))
```

```python
# Set operations — useful for comparing feature lists between datasets

train_features = {'age', 'income', 'credit_score', 'loan_amount', 'employed'}
test_features  = {'age', 'income', 'credit_score', 'loan_amount', 'region'}

print('In train but not test:', train_features - test_features)
print('In test but not train:', test_features - train_features)
print('In both (intersection):', train_features & test_features)
print('Combined (union):', train_features | test_features)
```

**Observation:** This set-difference check is exactly what you'd want when verifying that training and test datasets have the same columns before deploying a model.

## 5. Dictionaries — Key-Value Data

```python
# Dictionaries are used everywhere in ML:
# storing model metrics, hyperparameter configs, label mappings, etc.

model_metrics = {
    'accuracy':  0.834,
    'precision': 0.821,
    'recall':    0.849,
    'f1_score':  0.835
}

# Access
print('Accuracy:', model_metrics['accuracy'])

# .get() with a default — safer than direct access when key might not exist
print('AUC:', model_metrics.get('auc', 'not computed'))

# Iterating over key-value pairs
print('\nAll metrics:')
for metric, value in model_metrics.items():
    print(f'  {metric:12s}: {value:.3f}')
```

```python
# Dictionary comprehension — same idea as list comprehension
# Useful for encoding label mappings

categories = ['rock', 'mine', 'unknown']
label_encoding = {cat: idx for idx, cat in enumerate(categories)}
print('Label encoding:', label_encoding)

# Reverse it to decode predictions back to names
decode_map = {v: k for k, v in label_encoding.items()}
print('Decode map:', decode_map)

predicted_int = 1
print('Prediction decoded:', decode_map[predicted_int])
```

## 6. Control Flow

```python
# A function that categorises a model's accuracy
# Demonstrates if/elif/else with realistic ML thresholds

def evaluate_accuracy(acc):
    if acc >= 0.95:
        return 'Excellent — double-check for overfitting'
    elif acc >= 0.85:
        return 'Good'
    elif acc >= 0.70:
        return 'Acceptable — room for improvement'
    else:
        return 'Poor — revisit preprocessing and model choice'

for acc in [0.97, 0.88, 0.73, 0.58]:
    print(f'Accuracy {acc:.0%}: {evaluate_accuracy(acc)}')
```

```python
# enumerate() and zip() — two loop patterns used constantly in ML code

feature_names = ['age', 'income', 'credit_score']
feature_importances = [0.23, 0.45, 0.32]

print('Feature importances:')
for i, (name, importance) in enumerate(zip(feature_names, feature_importances)):
    print(f'  [{i}] {name:15s}: {importance:.2f}')
```

## 7. Functions

```python
# Writing a manual train-test split to understand what sklearn does internally
# This is NOT meant to replace sklearn — it is for understanding

import random

def simple_train_test_split(data, test_ratio=0.2, seed=42):
    """
    Split a list into training and test subsets.

    Parameters:
        data       : list of items to split
        test_ratio : fraction of data to use for testing (default 0.2)
        seed       : random seed for reproducibility (default 42)

    Returns:
        (train_data, test_data) as a tuple
    """
    random.seed(seed)
    data_copy = data.copy()     # avoid mutating the original
    random.shuffle(data_copy)

    n_test = int(len(data_copy) * test_ratio)
    test_data  = data_copy[:n_test]
    train_data = data_copy[n_test:]

    return train_data, test_data

# Test it
indices = list(range(20))   # simulating 20 data point indices
train, test = simple_train_test_split(indices, test_ratio=0.2)

print(f'Total: {len(indices)} | Train: {len(train)} | Test: {len(test)}')
print('Test indices:', sorted(test))
```

**Observation:** Writing this manually made `sklearn`'s `train_test_split` feel less like a magic function. The core steps are: shuffle, calculate split index, slice. The `stratify` parameter in sklearn adds an extra step — it makes sure each class appears in the same proportion in both splits — which this version does not do.

```python
# Demonstrating *args and **kwargs — appear in sklearn API signatures

def log_experiment(*metrics, **hyperparams):
    """
    Log an experiment's metrics and hyperparameters.
    *metrics    : any number of positional metric values
    **hyperparams: keyword arguments as hyperparameter settings
    """
    print('--- Experiment Log ---')
    print('Metrics:', metrics)
    print('Hyperparameters:')
    for k, v in hyperparams.items():
        print(f'  {k} = {v}')

log_experiment(0.834, 0.821, learning_rate=0.01, n_iterations=1000, regularization='none')
```

## 8. String Operations

```python
# String operations that appear frequently when processing column names or labels

raw_column_names = ['  Age ', 'Annual Income (k)', 'Credit Score', 'LOAN_AMOUNT', 'employed?']

def clean_column_name(name):
    """Standardize a column name to lowercase_with_underscores."""
    name = name.strip()              # remove leading/trailing whitespace
    name = name.lower()              # lowercase
    name = name.replace(' ', '_')    # spaces to underscores
    name = name.replace('(', '')     # remove parentheses
    name = name.replace(')', '')
    name = name.replace('?', '')     # remove question marks
    return name

cleaned = [clean_column_name(col) for col in raw_column_names]
print('Cleaned column names:')
for original, clean in zip(raw_column_names, cleaned):
    print(f'  {repr(original):35s} -> {clean}')
```

## 9. Experiment: List Lookup vs Set Lookup Speed

```python
# Sets use hash tables — O(1) lookup vs O(n) for lists
# This matters when checking whether a value belongs to a large collection

import time

large_list = list(range(1_000_000))
large_set  = set(large_list)

target = 999_999  # worst case for list — it's at the very end

# List lookup
start = time.time()
result_list = target in large_list
list_time = time.time() - start

# Set lookup
start = time.time()
result_set = target in large_set
set_time = time.time() - start

print(f'List lookup : {list_time*1000:.4f} ms  (found: {result_list})')
print(f'Set lookup  : {set_time*1000:.4f} ms  (found: {result_set})')
print(f'\nSet is approximately {list_time/max(set_time, 1e-9):.0f}x faster for this lookup')
```

**Observation:** The speed difference becomes significant when checking membership inside loops over large datasets. In ML preprocessing, if you need to check whether a value belongs to a list of allowed categories (e.g., filtering valid class labels), use a set rather than a list.
